### Python Packages & Libraries

```python
# Import the python packages & libraries 
import pandas as pd
from mlxtend.frequent_patterns import association_rules, apriori
from mlxtend.preprocessing import TransactionEncoder
```

### Read In Dataset

```python
# Utilize the Pandas library to create a dataframe (df) of our supplied dataset
df = pd.read_csv('medical_market_basket.csv')

# Preview the initial dataframe
pd.set_option("display.max_columns", None)
df.head()
```

### B2. Example Transaction

```python
# Example Transaction
example_transaction = df.dropna(how='all').iloc[1]  
example_transaction
```

### C1. Data Cleaning & Transformation Steps

#### Step #1. 
Check dataset column data types (dtype), value counts (non-null count) and total number of rows (entries) and columns using the info() method

```python
# Review dataset column data types, value counts and total rows/columns
df.info()
```

#### Step #2. 
Inspect and examine the newly created DataFrame (df) to identify any apparent high-level issues.

```python
# Remove empty rows
df_clean = df.dropna(how='all')  

# Preview the clean dataframe
df_clean.head()
```

#### Step #3. 
Create a list-of-lists (temp_list), where each inner list contains the items (prescriptions) associated with a single transaction (patient), showing a preview of the list by displaying the first 3 entries and last 3 entries. 

```python
# Convert the DataFrame to a list of lists, skipping NaN values
temp_list = [
    [str(item) for item in row if pd.notnull(item)]
    for row in df_clean.values
]

# Show first 3 list entries
for i in range(3):
    print(f"Index {i}: {temp_list[i]}")

print("...")

# Show last 3 list entries
for i in range(-3, 0):
    print(f"Index {len(temp_list) + i}: {temp_list[i]}")
```

#### Step #4. 
Encode the list of prescriptions (transaction) into one-hot encoded format. 

```python
# Encode the list of prescriptions (transactions) into one-hot encoded format
encoder = TransactionEncoder()
encoded_array = encoder.fit_transform(temp_list)

# Create a DataFrame from the encoded array
df_cleaned = pd.DataFrame(encoded_array, columns=encoder.columns_)

# Preview the one-hot encoded dataframe
df_cleaned.head()
```

#### Step #5. 
Double-check the transformed dataset column data types, value counts and total number of rows (entries) and columns using the info() method, again, to ensure it has been encoded properly (all of the dtypes, at this point, should be bool - True/False values).

```python
# Review transformed dataset column data types, value counts and total rows/columns
df_cleaned.info()
```

### Cleaned Data Set

```python
# Save the transformed dataset as a CSV file
df_cleaned.to_csv('d212_task3_transformed.csv', index=False)

print(f"The final cleaned and scaled dataset has been saved as:\n 'd212_task3_transformed.csv'")
```

### C2. Generating Association Rules with the Apriori Algorithm 

```python
# Generate frequent itemsets using the Apriori algorithm 
frequent_itemsets = apriori(df_cleaned, min_support=0.05, use_colnames=True)
frequent_itemsets
```

### C3. Association Rules Table

```python
# Generate association rules from the frequent itemsets
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.1)
rules
```

### C4. Relevant Rules Generated by the Apriori Algorithm

```python
# Display the top 3 rules (based on lift) 
top_rules = rules.sort_values(by='lift', ascending=False).head(3)
top_rules
```
